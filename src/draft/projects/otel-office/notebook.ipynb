# otel-office — usage walkthrough

Offline demonstration of the proof contract: deterministic simulator events drive the
office reducer, and an OTLP/JSON span maps onto a character state. The executable twin
lives in `examples/usage.ts` (gate-executed against `usage.expected.txt`).


```typescript
import { applyEvent, visibleCharacters } from "./src/model.js";
import { mapTracePayload } from "./src/otel.js";
import { createSimulation } from "./src/simulator.js";

const sim = createSimulation(7);
let state = sim.state;
for (let i = 0; i < 6; i += 1) for (const e of sim.step()) state = applyEvent(state, e);
console.log(`simulator: ${visibleCharacters(state, "station-local").length} characters after 6 ticks`);

const events = mapTracePayload({ resourceSpans: [] }, state);
console.log(`otlp: mapped ${events.length} event(s)`);
```
