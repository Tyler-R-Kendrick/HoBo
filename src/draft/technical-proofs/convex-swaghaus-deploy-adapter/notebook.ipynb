## Parse boundary

The executor response and the ambient env cross `src/parse.ts` once and
become refined domain types. The mock lane is marked, never live, and a
self-hosted receipt may not claim a `convex.cloud` host.

```typescript
import {
  parseDeployReceipt,
  shipSwaghausConvexApp,
} from "./src/index.ts";

const mocked = await shipSwaghausConvexApp({
  root: "examples/convex-swaghaus",
  deploymentName: "swaghaus",
  mock: true,
});
console.log(
  `${mocked.phase} mode=${mocked.phase === "deployed" ? mocked.receipt.mode : "-"} live=${mocked.preflight.live}`,
);

try {
  parseDeployReceipt({
    ok: true,
    deploymentUrl: "https://swaghaus.convex.cloud",
    deploymentName: "swaghaus",
    mocked: false,
    mode: "self-hosted",
  });
} catch (error) {
  console.log(`refused ${(error as Error).message}`);
}
```
